__Exercici 2:__ 

Volem saber quant valdria un producte si se li aplica un 10, 20, 30, 40 o 50 percent de descompte. L'import del producte l'ha de poder introduïr l'usuari.

```python
preu = input("Preu?")
try:
    preu = float(preu)
except ValueError:
    print("error")
else:
    for i in range(10, 60, 10):
        desc = preu * (i / 100)
        print(f"El preu final amb un {i}% de descompte és: {preu - desc}")
    
```

__Exercici 7:__
    
Calcular el factorial d'un nombre demanat per teclat. 

Exemple: 5! = 5 * 4 * 3 * 2 * 1

```python
num = input("Introdueix un nombre: ")
factorial = 0
try:
    num = int(num)
except ValueError:
    print("ERROR")
else:
    for i in range(num, 0, -1):
        factorial = factorial + (num * i)
        i -= 1
    print(f"Factorial = {factorial}") 
```
