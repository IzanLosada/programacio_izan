__Exercici 1:__
 
Introduir un número i visualitzar si és positiu, negatiu o neutre.

__Exercici 4:__
    
Introduir 4 números i dir si la suma dels dos primers és més gran, igual o més petit que la suma dels 2 següents.


```python
n1 = int(input("Introdueix un numero"))
n2 = int(input("Introdueix un numero"))
n3 = int(input("Introdueix un numero"))
n4 = int(input("Introdueix un numero"))

if (n1 + n2) > (n3 + n4):
    print("Son més grans els primers")
elif (n1 + n2) == (n3 + n4):
    print("Son iguals")
else:
    print("Son més grans els ultims")
```

__Exercici 6:__ 
    
Introduir 3 números i a continuació que els escrigui ordenats de major a menor.

```python
n1 = int(input("Introdueix un numero"))
n2 = int(input("Introdueix un numero"))
n3 = int(input("Introdueix un numero"))

if n1 >= n2 and n1 >= n3:
    if n2 >= n3:
        print(n1, n2, n3)
    else:
        print(n1, n3, n2)
elif n2 >= n1 and n2 >= n3:
    if n1 >= n3:
        print(n2, n1, n3)
    else:
        print(n2, n3, n1)
else:
    if n1 >= n2:
        print(n3, n1, n2)
    else:
        print(n3, n2, n1)
```

```python
c1 = float(input("Introdueix un costat"))
c2 = float(input("Introdueix un costat"))
d = input("Introdueix A/P").upper()

if d == "A":
    print(f"Àrea: {c1 * c2}")
elif d == "P":
    print(f"Perímetre: {(c1 * c1) + (c2 * c2)}")
```
